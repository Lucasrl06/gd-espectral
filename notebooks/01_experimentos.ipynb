# Experimentos do Projeto 3

Os fatores são variados separadamente. Para gerar os arquivos, executar `python -m experiments.reproduzir` na raiz do projeto. Este notebook lê os resultados; não substitui a execução do algoritmo.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Image

raiz = Path.cwd()
if raiz.name == 'notebooks':
    raiz = raiz.parent
dados = pd.read_csv(raiz / 'experiments/resultados/execucoes.csv')
print('Total:', len(dados))
display(pd.crosstab(dados.passo, dados.estado))
```

## Condicionamento

Comparação dos três pré-processamentos. Na experiência de correlação, usar o valor amostral, que pode diferir do valor alvo. Cada linha abaixo representa um cenário, não uma nova inicialização.

```python
colunas = ['fator', 'nivel', 'pre', 'rho_amostral', 'kappa']
display(dados[colunas].drop_duplicates().round(5))
display(Image(filename=str(raiz / 'figures/condicionamento.png')))
```

## Iterações previstas e observadas

A previsão corresponde à direção de pior caso. A comparação usa o passo 1/L, como em H7.1. Casos com limite de iterações não têm uma contagem observada até a tolerância.

```python
pior = dados[(dados.passo == '1/L') & (dados.tipo_inicializacao == 'pior_caso')]
display(pior[['fator', 'nivel', 'pre', 'kappa', 'iter_prevista', 'iter_observada', 'razao_obs_prev', 'estado']])
display(Image(filename=str(raiz / 'figures/observado_vs_previsto.png')))
```
